```python
import heapq

def dijkstra(graph, source):
    """
    Dijkstra's Algorithm with Hop Count
    Time: O((V + E) log V), Space: O(V)
    graph: dict {u: [(v, weight), ...]}, 0-indexed
    """
    n = len(graph)
    dist = [float('inf')] * n
    hops = [float('inf')] * n
    prev = [None] * n

    dist[source] = 0
    hops[source] = 0

    pq = [(0, 0, source)]  # (distance, hops, vertex)
    visited = set()

    while pq:
        d, h, u = heapq.heappop(pq)

        if u in visited:
            continue
        visited.add(u)

        for v, w in graph[u]:
            new_dist = dist[u] + w
            new_hops = hops[u] + 1

            if (new_dist < dist[v]) or (new_dist == dist[v] and new_hops < hops[v]):
                dist[v] = new_dist
                hops[v] = new_hops
                prev[v] = u
                heapq.heappush(pq, (dist[v], hops[v], v))

    return dist, hops, prev


def reconstruct_path(prev, source, target):
    path = []
    node = target

    while node is not None:
        path.append(node)
        node = prev[node]

    path.reverse()

    if path[0] == source:
        return path
    return []


# --- Graph with 6 Vertices ---
graph = {
    0: [(1, 2), (2, 2)],
    1: [(3, 2), (4, 5)],
    2: [(4, 2)],
    3: [(5, 2)],
    4: [(5, 2)],
    5: []
}

source = 0

dist, hops, prev = dijkstra(graph, source)

print(f'Shortest paths from vertex {source}:')
print(f'{"Vertex":>8} {"Distance":>10} {"Hops":>8} {"Path":>30}')
print('-' * 65)

for v in range(len(graph)):
    path = reconstruct_path(prev, source, v)
    path_str = ' -> '.join(map(str, path)) if path else 'No path'
    d = dist[v] if dist[v] != float('inf') else 'INF'
    h = hops[v] if hops[v] != float('inf') else 'INF'
    print(f'{v:>8} {str(d):>10} {str(h):>8} {path_str:>30}')
```
