```python
import random
comparison_count = 0  # Global counter
def min_max_dc(arr, low, high):
    global comparison_count
    # Base case: single element
    if low == high:
        return arr[low], arr[low]
    # Base case: two elements
    if high == low + 1:
        comparison_count += 1
        if arr[low] < arr[high]:
            return arr[low], arr[high]
        return arr[high], arr[low]
    # Divide
    mid = (low + high) // 2
    lmin, lmax = min_max_dc(arr, low, mid)
    rmin, rmax = min_max_dc(arr, mid + 1, high)
    # Conquer: combine with 2 comparisons
    comparison_count += 1
    overall_min = lmin if lmin < rmin else rmin
    comparison_count += 1
    overall_max = lmax if lmax > rmax else rmax
    return overall_min, overall_max
def min_max_naive(arr):
    mn, mx = arr[0], arr[0]
    comps = 0
    for x in arr[1:]:
        comps += 1
        if x < mn: mn = x
        comps += 1
        if x > mx: mx = x
    return mn, mx, comps
# --- Demonstration on small array --
arr = [3, 1, 7, 4, 9, 2, 8, 5, 6, 0]
comparison_count = 0
mn, mx = min_max_dc(arr, 0, len(arr) - 1)
dc_comps = comparison_count
_, _, naive_comps = min_max_naive(arr)
print(f'Array: {arr}')
print(f'Min: {mn}, Max: {mx}')
print(f'D&C Comparisons: {dc_comps}')
print(f'Naive Comparisons: {naive_comps}')
# --- Performance Analysis --
print(f'\n{"Size":>8} {"DC Comps":>12} {"Naive Comps":>14} {"Formula 3n/2-2":>16}')
print('-' * 56)
for size in [10, 100, 1000, 10000]:
    arr = [random.randint(1, 10000) for _ in range(size)]
    comparison_count = 0
    mn, mx = min_max_dc(arr, 0, len(arr) - 1)
    dc = comparison_count
    _, _, naive = min_max_naive(arr)
    formula = 3 * size // 2 - 2
    print(f'{size:>8} {dc:>12} {naive:>14} {formula:>16}')
```

```python
import random

comparison_count = 0  # Global counter


def min_max_second_dc(arr, low, high):
    global comparison_count

    # Base case: single element
    if low == high:
        return arr[low], None, arr[low], None

    # Base case: two elements
    if high == low + 1:
        comparison_count += 1
        if arr[low] < arr[high]:
            return arr[low], arr[high], arr[high], arr[low]
        else:
            return arr[high], arr[low], arr[low], arr[high]

    # Divide
    mid = (low + high) // 2

    lmin, lsecmin, lmax, lsecmax = min_max_second_dc(arr, low, mid)
    rmin, rsecmin, rmax, rsecmax = min_max_second_dc(arr, mid + 1, high)

    # Combine Minimum
    comparison_count += 1
    if lmin < rmin:
        overall_min = lmin
        candidates = [x for x in [lsecmin, rmin] if x is not None]
        comparison_count += 1
        overall_secmin = min(candidates)
    else:
        overall_min = rmin
        candidates = [x for x in [rsecmin, lmin] if x is not None]
        comparison_count += 1
        overall_secmin = min(candidates)

    # Combine Maximum
    comparison_count += 1
    if lmax > rmax:
        overall_max = lmax
        candidates = [x for x in [lsecmax, rmax] if x is not None]
        comparison_count += 1
        overall_secmax = max(candidates)
    else:
        overall_max = rmax
        candidates = [x for x in [rsecmax, lmax] if x is not None]
        comparison_count += 1
        overall_secmax = max(candidates)

    return overall_min, overall_secmin, overall_max, overall_secmax


def min_max_naive(arr):
    comps = 0

    s = sorted(arr)
    comps += len(arr) * (len(arr) - 1) // 2   # Approximate comparisons

    return s[0], s[1], s[-1], s[-2], comps


# --- Demonstration on Small Array ---
arr = [3, 1, 7, 4, 9, 2, 8, 5, 6, 0]

comparison_count = 0
mn, secmn, mx, secmx = min_max_second_dc(arr, 0, len(arr) - 1)
dc_comps = comparison_count

_, _, _, _, naive_comps = min_max_naive(arr)

print(f'Array: {arr}')
print(f'Min: {mn}')
print(f'Second Min: {secmn}')
print(f'Max: {mx}')
print(f'Second Max: {secmx}')
print(f'D&C Comparisons: {dc_comps}')
print(f'Naive Comparisons (Approx.): {naive_comps}')

# --- Performance Analysis ---
print(f'\n{"Size":>8} {"DC Comps":>12} {"Naive Comps":>16}')
print('-' * 40)

for size in [10, 100, 1000, 10000]:
    arr = [random.randint(1, 100000) for _ in range(size)]

    comparison_count = 0
    min_max_second_dc(arr, 0, len(arr) - 1)
    dc = comparison_count

    _, _, _, _, naive = min_max_naive(arr)

    print(f'{size:>8} {dc:>12} {naive:>16}')

print("\nAnalysis:")
print("Finding second minimum and second maximum requires")
print("additional comparisons during the combine step.")
print("Compared to finding only the minimum and maximum,")
print("approximately 2 extra comparisons are performed")
print("at each merge level of the recursion tree.")
```

```python
import random

comparison_count = 0  # Global counter


def partition(arr, low, high):
    global comparison_count

    pivot = arr[high]
    i = low

    for j in range(low, high):
        comparison_count += 1
        if arr[j] <= pivot:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1

    arr[i], arr[high] = arr[high], arr[i]
    return i


def quick_select(arr, low, high, k):
    if low == high:
        return arr[low]

    pivot_index = partition(arr, low, high)

    if pivot_index == k:
        return arr[pivot_index]
    elif k < pivot_index:
        return quick_select(arr, low, pivot_index - 1, k)
    else:
        return quick_select(arr, pivot_index + 1, high, k)


# --- Demonstration on Small Array ---
arr = [12, 3, 5, 7, 19, 26, 4, 15]

print(f"Array: {arr}")

comparison_count = 0
minimum = quick_select(arr.copy(), 0, len(arr) - 1, 0)
min_comp = comparison_count

comparison_count = 0
maximum = quick_select(arr.copy(), 0, len(arr) - 1, len(arr) - 1)
max_comp = comparison_count

comparison_count = 0
median = quick_select(arr.copy(), 0, len(arr) - 1, len(arr) // 2)
med_comp = comparison_count

print(f"Minimum Element : {minimum}")
print(f"Maximum Element : {maximum}")
print(f"Median Element  : {median}")

# --- Performance Test on 1000 Random Integers ---
size = 1000
arr = [random.randint(1, 10000) for _ in range(size)]

print(f"\nPerformance Test (Array Size = {size})")
print("-" * 50)

comparison_count = 0
minimum = quick_select(arr.copy(), 0, size - 1, 0)
min_comp = comparison_count

comparison_count = 0
maximum = quick_select(arr.copy(), 0, size - 1, size - 1)
max_comp = comparison_count

comparison_count = 0
median = quick_select(arr.copy(), 0, size - 1, size // 2)
med_comp = comparison_count

print(f'{"k":>10} {"Element":>12} {"Comparisons":>15}')
print("-" * 40)
print(f'{"1 (Min)":>10} {minimum:>12} {min_comp:>15}')
print(f'{"n (Max)":>10} {maximum:>12} {max_comp:>15}')
print(f'{"n//2":>10} {median:>12} {med_comp:>15}')

print("\nAnalysis:")
print("QuickSelect is a Divide and Conquer algorithm.")
print("Average Time Complexity : O(n)")
print("Worst Case Complexity   : O(n^2)")
print("Space Complexity        : O(1) (excluding recursion stack)")
```
